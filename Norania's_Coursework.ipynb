```python
#%% 2c)
from math import sqrt
import pandas as pd
data = pd.read_csv("https://gist.githubusercontent.com/simonlast/d20a53af004bfd7cd912403884ce80d2/raw/c92846e3387f11abc7e318db88c7aca6ecec1b4c/height_weight_data.csv")
data.columns = data.columns.str.strip()
height = data["Height (Inches)"]

n=len(height)
mean_height=height.mean()
std_height=height.std()
z= 1.645
se=std_height / sqrt(n)
lower = mean_height - z * se
upper = mean_height + z * se
print("Confidence interval for 90%: ", (lower, upper))
```

```python
#%% 3a)
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

data = pd.read_csv("https://raw.githubusercontent.com/tarunramavath/Salary-Prediction-Using-Linear-Regression/main/Salary_Data.csv")
X=data.iloc[:,:-1]
Y=data.iloc[:,-1]
model=LinearRegression().fit(X,Y)
slope =model.coef_.flatten()
intercept =model.intercept_

print(f"Slope: {slope}")
print(f"Intercept: {intercept}")

Y_predict=model.predict(X)

plt.scatter(X,Y,color="red", label="Actual data")
plt.plot(X,model.predict(X), label="Regression Line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.legend()
plt.show()

X_new=[[2.5]]
Y_new = model.predict(X_new)
print(f"Predicted salary for 2.5 years of experience: {Y_new[0]}")

X_new = [[13.5]]
Y_new = model.predict(X_new)
print(f"Predicted salary for 13.5 years of experience: {Y_new[0]}")
```

```python
#%% 3b)
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
data = pd.read_csv("https://raw.githubusercontent.com/tarunramavath/Salary-Prediction-Using-Linear-Regression/main/Salary_Data.csv")
X=data.iloc[:,:-1]
Y=data.iloc[:,-1]
model=LinearRegression().fit(X,Y)
Y_predict=model.predict(X)
r2 = r2_score(Y,Y_predict)
mse=mean_squared_error(Y,Y_predict)
print(f"R^2: {r2}")
print(f"Mean Squared Error: {mse} ")
```

```python
#%% 3C)
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split

data = pd.read_csv("https://raw.githubusercontent.com/tarunramavath/Salary-Prediction-Using-Linear-Regression/main/Salary_Data.csv")
X = data.iloc[:, :-1]
Y = data.iloc[:, -1]

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.25, random_state=23)

fit=LinearRegression().fit(X_train,Y_train)
slope =fit.coef_.flatten()[0]
intercept =fit.intercept_
print(f"Slope: {slope} ")
print(f"Intercept: {intercept}")

test_r2 = fit.score(X_test, Y_test)
train_r2= fit.score(X_train, Y_train)

print(f"R^2 for training data is: {train_r2}")
print(f"R^2 for testing data is: {test_r2}")

X_new = [[2.5]]
Y_new = fit.predict(X_new)
print(f"Predicted salary for 2.5 years of experience: {Y_new[0]}")

X_new = [[13.5]]
Y_new = fit.predict(X_new)
print(f"Predicted salary for 13.5 years of experience: {Y_new[0]}")
```

```python
#%% a)
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
data= pd.read_csv("https://raw.githubusercontent.com/dahjan/Diabetes-Dataset--Analysis/master/diabetes.csv")
X=data[["Glucose"]]
Y=data["Outcome"]

model = LogisticRegression(penalty=None, max_iter=10000).fit(X,Y)

b0=model.intercept_.flatten()
b1=model.coef_.flatten()
print(f"Intercept: {b0}")
print(f"Slope: {b1}")
print(f"y=1/(1+exp(-({b0}+{b1}x)))")

x_plot = np.linspace(X["Glucose"].min(), X["Glucose"].max(), 200)
p_plot = 1 / (1 + np.exp(-(b0 + b1 * x_plot)))

plt.plot(x_plot, p_plot, color='red', label='Logistic function')
plt.scatter(X["Glucose"], Y, color='blue', alpha=0.4, label='Data Points')
plt.title("Logistic Regression (Glucose -> Outcome)") # Corrected from plt.tile to plt.title
plt.xlabel("Glucose")
plt.ylabel("Probability of Outcome")
plt.legend()
plt.show()

patients = pd.DataFrame({'Glucose': [142,93,109,140]})
pred_class = model.predict(patients)
pred_prob = model.predict_proba(patients)[:,1]

print(f"Predicted Class:{pred_class}")
print(f"Predicted Probabilities:{pred_prob}")
```

```python
#%% b)
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
data = pd.read_csv("https://raw.githubusercontent.com/dahjan/Diabetes-Dataset--Analysis/master/diabetes.csv")

X_multi=data.iloc[:, [1,2,4,7]]
Y= data.iloc[:,-1]

model_multivariable= LogisticRegression(penalty=None, max_iter=20000)
model_multivariable.fit(X_multi,Y)

b0=model_multivariable.intercept_.flatten()
b1=model_multivariable.coef_.flatten()

print(f"Intercept: {b0}")
print(f"Coefficients: {b1}")

patients = pd.DataFrame([[142,82,64,21],[93,50,64,23],[109,64,99,26],[140,82,325,58]], columns=["Glucose", "BloodPressure", "Insulin", "Age"])

pred_class = model_multivariable.predict(patients)
pred_prob = model_multivariable.predict_proba(patients)[:,1]
print(f"Predicted Class:{pred_class}")
print(f"Predicted Probabilities:{pred_prob}")
```
